```python
# Import necessary libraries
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
# Set random seed for reproducibility
torch.manual_seed(42)
#Neural networks start with randomly initialized weights and biases.
#The seed ensures that this randomness is repeatable.
#Changing the seed will result in a different set of starting weights and biases for each layer of the model.
#This can lead to different training dynamics and possibly different final performance (e.g., loss and accuracy).
#With torch.manual_seed(42), every run of your script will produce the same results, provided all other factors remain constant.
```

```python
# Define a simple neural network for predicting exam scores
class ExamScoreNN(nn.Module):
    def __init__(self):
        super(ExamScoreNN, self).__init__()
        # Define a single hidden layer with 5 neurons
        #YOUR CODE GOES HERE
        self.hidden = nn.Linear(2,5)

        # Define the output layer with 1 neuron
        #YOUR CODE GOES HERE
        self.output = nn.Linear(5,1)

    def forward(self, x):
        # Apply ReLU activation to the hidden layer
        x = torch.relu(self.hidden(x))
        # Pass the result through the output layer
        x = self.output(x)
        return x
```

```python
# Instantiate the neural network
model = ExamScoreNN()

# Print the model architecture
print("Model Architecture:")
print(model)
```

```python
# Define a simple dataset for predicting exam scores
# Features: [Hours Studied, Hours of Sleep Before Exam]
data = torch.tensor([
    [1.0, 5.0],  # Example: 1 hour studied, 5 hours of sleep
    [2.0, 6.0],  # Example: 2 hours studied, 6 hours of sleep
    [3.0, 4.0],  # Example: 3 hours studied, 4 hours of sleep
    [4.0, 7.0]   # Example: 4 hours studied, 7 hours of sleep
])
# Labels: [Exam Scores]
labels = torch.tensor([
    [50.0],  # Predicted exam score for 1 hour studied, 5 hours sleep
    [65.0],  # Predicted exam score for 2 hours studied, 6 hours sleep
    [70.0],  # Predicted exam score for 3 hours studied, 4 hours sleep
    [85.0]   # Predicted exam score for 4 hours studied, 7 hours sleep
])
```

```python
# Define the loss function (Mean Squared Error)
criterion = nn.MSELoss()

# Define the optimizer (Stochastic Gradient Descent)
optimizer = optim.SGD(model.parameters(), lr=0.01)

# Number of training epochs
epochs = 100
```

```python
# Training loop
for epoch in range(epochs):
    # Forward pass
    outputs = model(data)
    loss = criterion(outputs, labels)

    # Backpropagation
    optimizer.zero_grad()  # Clear previous gradients
    loss.backward()        # Compute gradients
    optimizer.step()       # Update weights

    # Print loss every 100 epochs
    if (epoch + 1) % 100 == 0:
        print(f"Epoch [{epoch + 1}/{epochs}], Loss: {loss.item():.4f}")
```

```python
# Test the trained model
with torch.no_grad():
    test_outputs = model(data)
    print("\nFinal Predictions:")
    print(test_outputs)
    print("Actual Labels:")
    print(labels)
```
